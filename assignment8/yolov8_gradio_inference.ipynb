# YOLOv8 Object Detection with Gradio

This notebook demonstrates how to:
1. Install the Ultralytics YOLOv8 package
2. Load a pre-trained YOLOv8m model
3. Create an interactive Gradio interface for object detection
4. Perform inference on uploaded images

## 1. Install Required Packages

We'll install:
- `ultralytics`: The official YOLOv8 package
- `gradio`: For creating the interactive web interface

```python
#pip install ultralytics gradio pillow
```

## 2. Import Libraries

Import the necessary libraries for YOLOv8 and Gradio

```python
from ultralytics import YOLO
import gradio as gr
from PIL import Image
import cv2
import numpy as np
```

## 3. Load YOLOv8 Model

Load the pre-trained YOLOv8m model. The model will be automatically downloaded on first use.

**Model variants:**
- `yolov8n.pt` - Nano (fastest, least accurate)
- `yolov8s.pt` - Small
- `yolov8m.pt` - Medium (balanced) ✓ Using this
- `yolov8l.pt` - Large
- `yolov8x.pt` - Extra Large (most accurate, slowest)

```python
# Load YOLOv8m model (will download automatically if not present)
model = YOLO('yolov8m.pt')
print("YOLOv8m model loaded successfully!")
```

## 4. Create Inference Function

Define a function that takes an image and returns the detection results with bounding boxes

```python
def detect_objects(image, confidence_threshold=0.25):
    """
    Perform object detection on the input image
    
    Args:
        image: PIL Image or numpy array
        confidence_threshold: Minimum confidence for detections (0-1)
    
    Returns:
        Annotated image with bounding boxes and labels
    """
    # Run inference
    results = model(image, conf=confidence_threshold)
    
    # Get the annotated image
    annotated_image = results[0].plot()
    
    # Convert from BGR to RGB (OpenCV uses BGR, but Gradio expects RGB)
    annotated_image = cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB)
    
    return annotated_image
```

## 5. Create Gradio Interface with Webcam Support

Build an interactive web interface using Gradio with webcam support for live inference

```python
# Create Gradio interface with webcam support
interface = gr.Interface(
    fn=detect_objects,
    inputs=[
        gr.Image(sources=["webcam", "upload"], type="pil", label="Upload Image or Use Webcam"),
        gr.Slider(minimum=0.1, maximum=1.0, value=0.25, step=0.05, 
                 label="Confidence Threshold")
    ],
    outputs=gr.Image(type="numpy", label="Detection Results"),
    title="🎯 YOLOv8m Object Detection with Live Webcam",
    description="Upload an image or use your webcam for live object detection using YOLOv8m. Adjust the confidence threshold to filter detections.",
    theme="default",
    live=False  # Set to True for continuous webcam processing (may be slow)
)

# Launch the interface
interface.launch(share=True, debug=True)
```
